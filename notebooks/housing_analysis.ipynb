# 04 Cleaning the Data

For cleaning the data I first convert the byte objects into normal strings. I also check if there any missing values.

```python
# Convert byte objects into normal strings

for column in df.select_dtypes(["object"]):
    df[column] = df[column].str.decode("utf-8")


df.isna().sum()
```
